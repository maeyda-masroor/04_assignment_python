```python
import random

def play():
  user = input("what ur choice r for rock , p for paper and s for sissor")
  computer = random.choice(['r','s','p'])
  if computer == user :
    return 'tie'
  if is_win(user,computer):
    return "u won"
  return "u loss"
  #r>s s>p ,p>r

def is_win(player,opponent):
  #return true if players wins
  if (player == 'r' and opponent == 's') or (player == 's' and opponent == 'p') or (player == 'p' and opponent == 'r'):
    return True

print(play())
```
